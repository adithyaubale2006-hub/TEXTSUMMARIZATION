# Text Summarization

This notebook demonstrates a simple text summarization example using the Hugging Face Transformers `pipeline`.

Instructions:
1. Install dependencies (the cell below uses pip).
2. Run the summarization pipeline on your text.


```python
# Install required packages (uncomment and run if needed)
# !pip install -q transformers[sentencepiece] datasets

from transformers import pipeline

# Create a summarization pipeline (this will download a default model)
summarizer = pipeline('summarization')

# Replace the text below with the document you want to summarize
text = '''
Replace this string with the long text you want to summarize.
This is a placeholder example sentence intended to show how to call the summarizer.
'''

# Generate summary
summary = summarizer(text, max_length=130, min_length=30, do_sample=False)
print(summary[0]['summary_text'])
```
